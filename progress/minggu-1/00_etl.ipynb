## Sumber data dan strategi ekstraksi

Tiap file bulanan berisi satu sheet `REKAPITULASI` (matriks komoditas x tanggal) dan satu sheet per hari (matriks komoditas x 7 pasar).

**Keputusan: sheet harian jadi sumber utama.** Alasannya, hasil audit menemukan:

1. Kolom hari pada `REKAPITULASI` tidak konsisten posisinya antar file (hari ke-1 kadang di kolom indeks 2, kadang 3), sehingga pembacaan dengan offset tetap akan bergeser satu hari di 34 dari 44 file.
2. `REKAPITULASI` menyimpang dari rata-rata 7 pasar pada puluhan sel (contoh: bawang merah 24 Maret 2023 tercatat 5.571 padahal rata-rata 7 pasar 33.143 — 5.571 adalah nilai total dibagi 7).
3. Sheet harian memuat data 7 pasar secara eksplisit, jadi rata-ratanya dapat diverifikasi ulang.

`REKAPITULASI` tetap dibaca sebagai pembanding dan sebagai cadangan bila sheet harian tidak tersedia. Setiap selisih dicatat ke `data/etl_konflik.csv`.

```python
# ==================================================================
# KONSTANTA
# ==================================================================
PASAR_KEYWORD = "pasar"      # penanda kolom pasar pada sheet harian
RATA_KEYWORD = "rata"        # penanda kolom rata-rata (dihitung ulang, tidak dipakai)

# Pola pencocokan nama komoditas. Urutan = prioritas.
# Catatan: Jan-Feb 2023 belum memakai istilah "Premium"; kualitas tertinggi
# saat itu ditulis "Beras IR.64 Kwalitas I".
COMMODITY_PATTERNS = {
    "cabe_rawit": [
        lambda n: "rawit" in n,
    ],
    "bawang_merah": [
        lambda n: "bawang merah" in n and "lokal" in n,
        lambda n: "bawang merah" in n and "bombay" not in n and "putih" not in n,
    ],
    "beras_premium": [
        lambda n: "beras" in n and "premium" in n,
        lambda n: "beras ir.64" in n and "kwalitas i" in n and "kwalitas ii" not in n,
    ],
}

BULAN_ID = {
    "januari": 1, "februari": 2, "maret": 3, "april": 4, "mei": 5, "juni": 6,
    "juli": 7, "agustus": 8, "september": 9, "oktober": 10, "november": 11,
    "desember": 12,
}


# ==================================================================
# UTILITAS
# ==================================================================
def to_number(value):
    """
    Ubah isi sel jadi float > 0, atau None bila bukan harga valid.

    Menangani: None, sel kosong, '  ', '-', '#REF!', '#N/A', angka 0,
    angka negatif, dan string berformat ribuan.
    """
    if value is None or isinstance(value, bool):
        return None
    if isinstance(value, (int, float)):
        v = float(value)
        return v if np.isfinite(v) and v > 0 else None
    if isinstance(value, str):
        s = value.strip().replace(",", "")
        if s == "" or s in {"-", "–", "—"}:
            return None
        if s.startswith("#"):          # #REF!, #N/A, #DIV/0!
            return None
        try:
            v = float(s)
        except ValueError:
            return None
        return v if np.isfinite(v) and v > 0 else None
    return None


def find_daily_header(rows, max_scan=14):
    """
    Cari baris header sheet harian: harus punya kolom 'Pasar ...' dan 'Rata-Rata'.
    Mengembalikan baris header atau None.
    """
    for row in rows[:max_scan]:
        if not row:
            continue
        has_pasar = any(v and PASAR_KEYWORD in str(v).lower() for v in row)
        has_rata = any(v and RATA_KEYWORD in str(v).lower() for v in row)
        if has_pasar and has_rata:
            return row
    return None


def find_day_columns(rows, max_scan=14):
    """
    Cari baris header REKAPITULASI berisi nomor hari 1..31.
    Kembalikan {hari: indeks_kolom}.

    Kenapa tidak boleh `col_idx = 2 + day`?
    Posisi kolom hari ke-1 tidak konsisten: Jan-Jun 2023 di indeks 3,
    Jul-Des 2023 dan 2024-2026 di indeks 2. Sebagian file juga melompat
    nomornya (April 2023: ...21, 24, 25...).
    """
    best = None
    for row in rows[:max_scan]:
        if not row:
            continue
        mapping = {}
        for idx, val in enumerate(row):
            if isinstance(val, bool):
                continue
            if isinstance(val, (int, float)) and float(val).is_integer():
                day = int(val)
                if 1 <= day <= 31 and day not in mapping:
                    mapping[day] = idx
        if len(mapping) >= 20 and 1 in mapping:
            if best is None or len(mapping) > len(best):
                best = mapping
    return best


def find_commodity_row(rows, key):
    """Cari baris komoditas pada daftar baris sheet. Kembalikan baris atau None."""
    for row in rows:
        if not row or len(row) < 2 or row[1] is None:
            continue
        name = str(row[1]).lower().strip()
        if any(p(name) for p in COMMODITY_PATTERNS[key]):
            return row
    return None


def read_sheet_date(rows, max_scan=8):
    """
    Baca tanggal yang tertulis di dalam sheet ('Hari / Tanggal : Senin, 1 April 2023').
    Dipakai untuk mendeteksi label tanggal yang salah di file sumber.
    """
    for row in rows[:max_scan]:
        if not row:
            continue
        for v in row:
            if v and "tanggal" in str(v).lower():
                m = re.search(r"(\d{1,2})\s+([A-Za-z]+)\s+(\d{4})", str(v))
                if m:
                    bulan = BULAN_ID.get(m.group(2).lower())
                    if bulan:
                        return int(m.group(1)), bulan, int(m.group(3))
    return None


# ==================================================================
# EKSTRAKSI PER SHEET HARIAN  (SUMBER UTAMA)
# ==================================================================
def extract_daily_sheet(rows):
    """
    Hitung harga = rata-rata 7 pasar dari satu sheet harian.

    'Rata-Rata' milik file TIDAK dipakai: pada beberapa file nilainya
    tidak konsisten dengan rata-rata 7 kolom pasar. Kita hitung sendiri
    agar dapat diaudit.

    Kembalikan (dict komoditas->harga, info) atau (None, info).
    """
    header = find_daily_header(rows)
    if header is None:
        return None, {"status": "header_tidak_ditemukan", "pasar_terisi": {}}

    market_cols = [j for j, v in enumerate(header)
                   if v and PASAR_KEYWORD in str(v).lower()]
    if not market_cols:
        return None, {"status": "kolom_pasar_tidak_ada", "pasar_terisi": {}}

    result, filled = {}, {}
    for key in COMMODITY_PATTERNS:
        row = find_commodity_row(rows, key)
        if row is None:
            result[key] = np.nan
            filled[key] = 0
            continue
        values = [to_number(row[j]) for j in market_cols if j < len(row)]
        values = [v for v in values if v is not None]
        result[key] = float(np.mean(values)) if values else np.nan
        filled[key] = len(values)

    return result, {"status": "ok", "pasar_terisi": filled,
                    "jumlah_kolom_pasar": len(market_cols)}


# ==================================================================
# EKSTRAKSI SHEET REKAPITULASI  (PEMBANDING / FALLBACK)
# ==================================================================
def extract_rekap_sheet(rows):
    """Ambil nilai per hari dari sheet REKAPITULASI. Kembalikan ({hari: {kolom: nilai}}, info)."""
    day_cols = find_day_columns(rows)
    if not day_cols:
        return None, {"status": "header_hari_tidak_ditemukan"}

    commodity_rows = {k: find_commodity_row(rows, k) for k in COMMODITY_PATTERNS}
    per_day = {}
    for day, col_idx in day_cols.items():
        rec = {}
        for key, row in commodity_rows.items():
            raw = row[col_idx] if row is not None and col_idx < len(row) else None
            rec[key] = to_number(raw)
        per_day[day] = rec
    return per_day, {"status": "ok", "offset_hari_1": day_cols[1]}


# ==================================================================
# PEMROSESAN SATU FILE BULANAN
# ==================================================================
def process_month_file(file_path, year, month_num):
    """
    Proses satu file bulanan.

    Alur:
      1. Untuk setiap sheet harian (nama sheet = angka hari), hitung rata-rata 7 pasar.
      2. Baca sheet REKAPITULASI sebagai pembanding.
      3. Jika sheet harian tidak ada/kosong, pakai nilai REKAPITULASI.
      4. Catat setiap konflik ke daftar audit.

    Kembalikan (records, meta, conflicts).
    """
    fname = os.path.basename(file_path)
    days_in_month = calendar.monthrange(year, month_num)[1]

    meta = {
        "file": fname,
        "tahun": year,
        "bulan": month_num,
        "status": "ok",
        "offset_hari_1": None,
        "sheet_harian_terbaca": 0,
        "sheet_harian_kosong": [],
        "hari_tanpa_sumber": [],
        "tanggal_label_salah": [],
        "komoditas_tidak_ditemukan": [],
        "pasar_kurang_7": 0,
        "sumber": {},
    }
    conflicts = []

    try:
        wb = openpyxl.load_workbook(file_path, data_only=True, read_only=True)
    except Exception as e:
        meta["status"] = f"gagal_buka: {e}"
        return [], meta, conflicts

    sheet_names = list(wb.sheetnames)

    # ---------- 1. sheet harian ----------
    daily = {}
    for sn in sheet_names:
        if sn.strip().upper() == "REKAPITULASI":
            continue
        try:
            day = int(str(sn).strip())
        except (TypeError, ValueError):
            continue
        if not (1 <= day <= days_in_month):
            continue

        rows = list(wb[sn].iter_rows(values_only=True))
        values, info = extract_daily_sheet(rows)
        if values is None:
            meta["sheet_harian_kosong"].append(day)
            continue

        # cek label tanggal di dalam sheet
        label = read_sheet_date(rows)
        if label and label != (day, month_num, year):
            meta["tanggal_label_salah"].append(
                f"sheet {day} -> tertulis {label[0]}/{label[1]}/{label[2]}")

        # cek kelengkapan 7 pasar
        for key, n in info["pasar_terisi"].items():
            if n < 7:
                meta["pasar_kurang_7"] += 1

        daily[day] = values
        meta["sheet_harian_terbaca"] += 1

    # ---------- 2. sheet REKAPITULASI ----------
    rekap = {}
    if "REKAPITULASI" in sheet_names:
        rows = list(wb["REKAPITULASI"].iter_rows(values_only=True))
        rekap, info = extract_rekap_sheet(rows)
        if rekap is not None:
            meta["offset_hari_1"] = info["offset_hari_1"]
        else:
            rekap = {}
    wb.close()

    # ---------- 3. gabungkan ----------
    records = []
    for day in range(1, days_in_month + 1):
        rec = {"tanggal": date(year, month_num, day).strftime("%Y-%m-%d")}
        ada_data = False

        for key in COMMODITY_PATTERNS:
            val_daily = daily.get(day, {}).get(key, np.nan)
            val_rekap = rekap.get(day, {}).get(key, None)

            # nilai final: utamakan sheet harian
            if val_daily is not None and not (isinstance(val_daily, float) and np.isnan(val_daily)):
                final = val_daily
                meta["sumber"][key] = meta["sumber"].get(key, 0) + 1
                # bandingkan dengan REKAPITULASI bila keduanya ada
                if val_rekap is not None and abs(val_rekap - val_daily) > 1.0:
                    conflicts.append({
                        "tanggal": rec["tanggal"],
                        "komoditas": key,
                        "nilai_sheet_harian": round(float(val_daily), 2),
                        "nilai_rekapitulasi": round(float(val_rekap), 2),
                        "selisih": round(float(val_rekap - val_daily), 2),
                    })
            elif val_rekap is not None:
                final = val_rekap      # fallback
            else:
                final = np.nan

            rec[key] = final
            if not (isinstance(final, float) and np.isnan(final)):
                ada_data = True

        if ada_data:
            records.append(rec)
        else:
            meta["hari_tanpa_sumber"].append(day)

    # komoditas yang tidak pernah muncul di file ini
    all_rows = []
    meta["komoditas_tidak_ditemukan"] = [
        k for k in COMMODITY_PATTERNS
        if k not in meta["sumber"] and not rekap
    ]

    return records, meta, conflicts
```

## Eksekusi fungsi looping pada seluruh file

```python
all_data = []
etl_meta = []
etl_conflicts = []

for year, folder_name in YEAR_FOLDERS.items():
    folder_dir = os.path.join(RAW_DATA_PATH, folder_name)
    if not os.path.exists(folder_dir):
        print(f"[LEWAT] Direktori tidak ditemukan: {folder_dir}")
        continue

    file_list = sorted([
        f for f in os.listdir(folder_dir)
        if f.endswith(".xlsx") and not f.startswith("~$")
    ])
    print(f"\n=== Tahun {year} ({len(file_list)} file) ===")

    for fname in file_list:
        # Deteksi bulan dari PREFIX nama file, bukan substring.
        # Substring berbahaya: '02' in '03. REKAP MARET 2023.XLSX' -> True
        # karena tahun "2023" mengandung "02".
        m = re.match(r"^(\d{2})\.", fname)
        if not m:
            print(f"  [LEWAT] nomor bulan tidak terdeteksi: {fname}")
            continue
        month_num = int(m.group(1))
        if not 1 <= month_num <= 12:
            print(f"  [LEWAT] nomor bulan tidak valid ({month_num}): {fname}")
            continue

        file_path = os.path.join(folder_dir, fname)
        records, meta, conflicts = process_month_file(file_path, year, month_num)
        all_data.extend(records)
        etl_meta.append(meta)
        etl_conflicts.extend(conflicts)

        flag = []
        if meta["status"] != "ok":
            flag.append(meta["status"])
        if meta["sheet_harian_kosong"]:
            flag.append(f"sheet kosong: {meta['sheet_harian_kosong']}")
        if meta["hari_tanpa_sumber"]:
            flag.append(f"hari tanpa data: {meta['hari_tanpa_sumber']}")
        if meta["tanggal_label_salah"]:
            flag.append(f"label tanggal salah: {len(meta['tanggal_label_salah'])} sheet")
        if meta["pasar_kurang_7"]:
            flag.append(f"pasar<7: {meta['pasar_kurang_7']} sel")
        status = " | ".join(flag) if flag else "ok"

        print(f"  {fname[:44]:44s} {len(records):3d} rec  "
              f"harian={meta['sheet_harian_terbaca']:2d}  {status}")

print(f"\n{'='*66}")
print(f"Total record      : {len(all_data)}")
print(f"File diproses     : {len(etl_meta)}")
print(f"Konflik harian vs REKAPITULASI : {len(etl_conflicts)}")

# ---------- simpan log audit ----------
etl_log = pd.DataFrame(etl_meta)
for col in ["sheet_harian_kosong", "hari_tanpa_sumber", "tanggal_label_salah"]:
    etl_log[col] = etl_log[col].apply(lambda x: ",".join(map(str, x)))
etl_log["sumber"] = etl_log["sumber"].apply(
    lambda d: "; ".join(f"{k}={v}" for k, v in d.items()) if isinstance(d, dict) else "")
etl_log = etl_log[["tahun", "bulan", "file", "status", "sheet_harian_terbaca",
                   "sheet_harian_kosong", "hari_tanpa_sumber", "offset_hari_1",
                   "pasar_kurang_7", "tanggal_label_salah", "sumber"]]
etl_log.to_csv(os.path.join(DATA_DIR, "etl_log.csv"), index=False)

df_conflict = pd.DataFrame(etl_conflicts)
df_conflict.to_csv(os.path.join(DATA_DIR, "etl_konflik.csv"), index=False)

print(f"Log audit    -> {os.path.join(DATA_DIR, 'etl_log.csv')}")
print(f"Log konflik  -> {os.path.join(DATA_DIR, 'etl_konflik.csv')}")

if len(df_conflict):
    print("\nContoh konflik (REKAPITULASI menyimpang dari rata-rata 7 pasar):")
    print(df_conflict.head(10).to_string(index=False))
```

## Pembersihan grid tanggal, imputasi, dan interpolasi

```python
df_raw = pd.DataFrame(all_data)

# ===== 1. Validasi hasil ekstraksi =====
print(f"Record terkumpul    : {len(df_raw)}")
print(f"Tanggal duplikat    : {df_raw.duplicated(subset=['tanggal']).sum()}")

df_raw = (df_raw
          .drop_duplicates(subset=["tanggal"], keep="first")
          .sort_values("tanggal")
          .reset_index(drop=True))
df_raw["tanggal"] = pd.to_datetime(df_raw["tanggal"])

# ===== 2. Grid tanggal harian penuh =====
start, end = df_raw["tanggal"].min(), df_raw["tanggal"].max()
full_range = pd.date_range(start=start, end=end, freq="D")

df = (pd.DataFrame({"tanggal": full_range})
        .merge(df_raw, on="tanggal", how="left")
        .set_index("tanggal"))

print(f"\nRentang data        : {start.date()} s/d {end.date()}")
print(f"Panjang grid        : {len(df)} hari")
print(f"Record tersedia     : {df.notna().all(axis=1).sum()} hari")

# ===== 3. Tanggal yang tidak ada di file sumber =====
tanpa_record = df.index[df.isna().all(axis=1)]
print(f"\nTanggal tanpa record: {len(tanpa_record)}")
if len(tanpa_record):
    for d in tanpa_record:
        print(f"  - {d.strftime('%Y-%m-%d')}")

# ===== 4. Missing per kolom SEBELUM interpolasi =====
missing_before = df.isna().sum()
print("\nMissing per kolom SEBELUM interpolasi:")
print(missing_before)

# ===== 5. Interpolasi linear berbasis waktu =====
# Alasan: seri harian harga tidak boleh berlubang untuk SARIMA (m=7)
# dan untuk pembentukan fitur lag. Lubang maksimum hanya 3 hari berturut.
df = df.interpolate(method="time").bfill().ffill()

print("\nMissing SETELAH interpolasi (harus 0):")
print(df.isna().sum())
print(f"\nTotal baris kontinu : {len(df)} hari")

# ===== 6. Sanity check =====
print("\n=== Sanity check ===")
print(f"Nilai <= 0  : {(df[list(COMMODITY_PATTERNS)] <= 0).sum().to_dict()}")
print(f"Nilai NaN   : {df.isna().sum().to_dict()}")
print(f"Nilai inf   : {np.isinf(df[list(COMMODITY_PATTERNS)]).sum().to_dict()}")

print("\nStatistik ringkas:")
print(df.describe().T[["mean", "std", "min", "max"]].round(0).to_string())

print("\nCoefficient of Variation (CV):")
for c in COMMODITY_PATTERNS:
    cv = df[c].std() / df[c].mean()
    print(f"  {c:16s} = {cv:.4f}")

# ===== 7. Ringkasan mutu data =====
mutu = pd.DataFrame({
    "kolom": list(missing_before.index),
    "nilai_asli": len(df) - missing_before.values,
    "diinterpolasi": missing_before.values,
    "persen_interpolasi": (missing_before.values / len(df) * 100).round(2),
})
mutu.to_csv(os.path.join(DATA_DIR, "ringkasan_mutu_data.csv"), index=False)
print(f"\nRingkasan mutu -> {os.path.join(DATA_DIR, 'ringkasan_mutu_data.csv')}")
print(mutu.to_string(index=False))
```
